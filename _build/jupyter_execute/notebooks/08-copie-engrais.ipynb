# Ajout Fichier `Engrais.csv`

```python
import shutil
from pathlib import Path

# --- 1. CHEMINS ---
base_dir = Path.cwd().parent.resolve()

# Fichier source
source_path = base_dir / "data" / "Engrais" / "csv" / "raw" / "Engrais.csv"

# Fichier de destination final
destination_path = base_dir / "includes_sassemeV1" / "modeleAgricole" / "Engrais" / "Engrais.csv"


# --- 2. PROCESSUS DE COPIE ---
try:
    # S'assurer que le dossier de destination existe
    destination_path.parent.mkdir(parents=True, exist_ok=True)
    
    # Copier le fichier
    shutil.copy(source_path, destination_path)
    
    print("✅ Fichier 'Engrais.csv' copié avec succès.")
    print(f"   Source : {source_path}")
    print(f"   Destination : {destination_path}")

except FileNotFoundError:
    print(f"🚨 ERREUR : Le fichier source n'a pas été trouvé. Vérifiez le chemin :")
    print(f"   {source_path}")
except Exception as e:
    print(f"🚨 Une erreur est survenue : {e}")
```
