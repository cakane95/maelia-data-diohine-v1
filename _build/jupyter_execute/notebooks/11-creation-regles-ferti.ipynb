# Mise en place `reglesDeDecisions_fertilisation.csv`

```python
import pandas as pd
from pathlib import Path
```

```python
base_dir = Path.cwd().parent.resolve()

# Fichier en entrée
input_regles_path = base_dir / "data" / "ITK" / "csv" / "raw" / "reglesDeDecisions_fertilisation.csv"

# Fichier en sortie
output_ferti_path = base_dir / "includes_sassemeV1" / "modeleAgricole" / "culture" / "reglesDeDecisions_fertilisation.csv"

# Liste de nos 6 ITK cibles
itk_sasseme = [
    'arachide_precMil',
    'arachide_precJachere',
    'jachere_precMil',
    'jachere_precArachide',
    'mil_precArachide',
    'mil_precMil'
]
```

```python
try:
    df_source = pd.read_csv(input_regles_path, sep=';')
    print("✅ Fichier source des règles de fertilisation chargé avec succès.")
except Exception as e:
    print(f"🚨 ERREUR lors du chargement du fichier : {e}")
```

```python
# On garde les 2 premières colonnes de base et on ajoute nos 6 ITK
colonnes_base = df_source.columns[:2].tolist()
df_sasseme_ferti = df_source[colonnes_base].copy()

for itk in itk_sasseme:
    df_sasseme_ferti[itk] = "NA" # On initialise tout à NA par défaut
print("✅ Structure du nouveau DataFrame créée.")
```

```python
# Dictionnaire des paramètres fixes à appliquer
parametres_ferti = {
    'FERTIALT_NOM_ALTERNATIVE': 'fertilisation minerale',
    'FERTIALT_ORDRE_ALTERNATIVE': 1,
    'FERTIALT_ORDRE_APPORT': 1,
    'FERTIALT_NOM_PRODUIT': 'NPK',
    'FERTIALT_DOSE': 300,
    'FERTIALT_DOSE_P': 20,
    'FERTIALT_DOSE_K': 10,
    'FERTIALT_PROF_WSOL': 0,
    'FERTIALT_AGRIW': 'oui',
    'FERTIALT_OUTIL': 0,
    'FERTIALT_TPS_TRAVAIL': 0.0625, # 2 jours/ha converti en ha/h
    'FERTIALT_N_PASSAGES': 1,
    'FERTIALT_OT_SIMULTANEE': 'NA',
    'FERTIALT_N_SOUS_PERIODES': 1,
    'FERTIALT_DEBUT': 166, # 15 Juin
    'FERTIALT_FIN': 242   # 30 Août
}
```

```python
# Appliquer les valeurs à nos 6 colonnes d'ITK
for parametre, valeur in parametres_ferti.items():
    df_sasseme_ferti.loc[df_sasseme_ferti['FERTIALT_NOM_ITK'] == parametre, itk_sasseme] = valeur
print("✅ Paramètres de fertilisation spécifiques appliqués.")
```

```python
display(df_sasseme_ferti)
```

```python
output_ferti_path.parent.mkdir(parents=True, exist_ok=True)
df_sasseme_ferti.to_csv(output_ferti_path, sep=';', index=False)
print(f"\n✅ Fichier 'reglesDeDecisions_fertilisation.csv' sauvegardé dans :\n   {output_ferti_path}")
```
