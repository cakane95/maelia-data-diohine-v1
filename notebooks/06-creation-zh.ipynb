# Création du Fichier de la Zone Hydrographique (`ZH.shp`)

## 1. Objectif

Ce document décrit la création du fichier `ZH.shp`, qui représente la Zone Hydrographique (ZH) unique de la zone d'étude.

## 2. Méthodologie

Le processus consiste à :
1.  **Fusionner** l'ensemble des polygones du `parcellaire_enrichi.shp` en une seule entité géographique.
2.  **Créer** l'attribut requis `ID_ZH` avec une valeur fixe de `1`.

## 3. Fichiers en Entrée et en Sortie

* **Fichier en Entrée :** `data/sols/shapefiles/processed/parcellaire_enrichi.shp`
* **Fichier en Sortie :** `includes_sassemeV1/modeleHydrographique/zonesHydrographiques/ZH.shp`

## 4. Description des Attributs

| Variable | Description | Origine / Traitement |
| :--- | :--- | :--- |
| `ID_ZH` | Identifiant de la Zone Hydrographique. | Valeur fixe **1**. |

```python
import geopandas as gpd
from pathlib import Path

# --- 1. CHEMINS ---
base_dir = Path.cwd().parent.resolve()

# Fichier en entrée
input_parcellaire_path = base_dir / "data" / "sols" / "shapefiles" / "processed" / "parcellaire_enrichi.shp"

# Fichier en sortie
output_zh_path = base_dir / "includes_sassemeV1" / "modeleHydrographique" / "zonesHydrographiques" / "ZH.shp"


# --- 2. CHARGEMENT DES DONNÉES ---
try:
    gdf_parcellaire = gpd.read_file(input_parcellaire_path)
    print(f"✅ Parcellaire chargé avec {len(gdf_parcellaire)} polygones.")
except Exception as e:
    print(f"🚨 ERREUR lors du chargement du fichier : {e}")


# --- 3. FUSION DE TOUS LES POLYGONES ---
# La méthode .unary_union est la plus efficace pour tout fusionner en une seule géométrie
zone_hydro_poly = gdf_parcellaire.unary_union
print("-> Toutes les parcelles ont été fusionnées en un seul polygone.")


# --- 4. CRÉATION DU GEODATAFRAME FINAL ---
# On crée un nouveau GeoDataFrame avec une seule ligne et les attributs requis
gdf_zh = gpd.GeoDataFrame(
    {
        'ID_ZH': [1]
    },
    geometry=[zone_hydro_poly],
    crs=gdf_parcellaire.crs  # On conserve le CRS d'origine
)
print("✅ GeoDataFrame final pour la Zone Hydrographique créé.")


# --- 5. SAUVEGARDE ---
output_zh_path.parent.mkdir(parents=True, exist_ok=True)
gdf_zh.to_file(output_zh_path, driver='ESRI Shapefile', encoding='utf-8')
print(f"\n✅ Fichier 'ZH.shp' sauvegardé dans :\n   {output_zh_path}")


# --- 6. VÉRIFICATION ---
print("\nAperçu du fichier final :")
display(gdf_zh)
```
